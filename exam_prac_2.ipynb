```python
img="/content/drive/MyDrive/Dataset/pic/tom and jerry.png"
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img_bgr=cv2.imread(img)
img_rgb=cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB)
#display
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.axis('off')
plt.title('real')
plt.show()
```

```python
#rbga
img_rgba= cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGBA)
#display
plt.subplot(1,2,1)
plt.imshow(img_rgba,alpha=0.5)
plt.title("rgba")
plt.axis('off')
plt.show()
```

```python
#gray
gray=cv2.cvtColor(img_bgr,cv2.COLOR_BGR2GRAY)
plt.subplot(1,2,1)
plt.imshow(gray,cmap='gray')
plt.title('gray')
plt.axis('off')
plt.show()
```

```python
#binary
_,binary=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
plt.subplot(1,2,1)
plt.imshow(binary,cmap='gray')
plt.title('binary')
plt.axis('off')
plt.show()
```

```python
#neg
neg=255-gray
plt.subplot(1,2,1)
plt.imshow(neg,cmap='gray')
plt.title('neg')
plt.axis('off')
plt.show()
```

```python
#bright
bright=cv2.convertScaleAbs(gray,alpha=1.2,beta=50)
plt.subplot(1,2,1)
plt.imshow(bright,cmap='gray')
plt.title('bright')
plt.axis('off')
plt.show()
```

```python
#dark
dark=cv2.convertScaleAbs(gray,alpha=1.0,beta=50)
plt.subplot(1,2,1)
plt.imshow(dark,cmap='gray')
plt.title('dark')
plt.axis('off')
plt.show()
```

```python
#single
_,single=cv2.threshold(gray,120,255,cv2.THRESH_BINARY)
plt.subplot(1,2,1)
plt.imshow(single,cmap='gray')
plt.title('single')
plt.axis('off')
plt.show()
```

```python
#dual
dual=cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,11,2)
plt.subplot(1,2,1)
plt.imshow(dual,cmap='gray')
plt.title('dual')
plt.axis('off')
plt.show()
```

```python
#contrast
min_val= np.min(gray)
max_val= np.max(gray)
contrast = ((gray-min_val)/(max_val-min_val)*255).astype(np.uint8)
plt.imshow(contrast,cmap='gray')
plt.title('contrast')
plt.axis('off')
plt.show()
```

```python
#smooth
smooth=cv2.GaussianBlur(gray,(5,5),0)
plt.subplot(1,2,1)
plt.imshow(smooth,cmap='gray')
plt.title('smooth')
plt.axis('off')
plt.show()
```

```python
#sharp
sharp=np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
sharp_img=cv2.filter2D(gray,-1,sharp)
plt.imshow(sharp_img,cmap='gray')
```

```python
#hist
hist=cv2.calcHist([gray],[0],None,[256],[0,256])

plt.plot(hist)
plt.xlabel('pixel')
plt.ylabel('freq')
```

```python
#equal
eq=cv2.equalizeHist(gray)
plt.imshow(eq,cmap='gray')
```

```python
#median
median=cv2.medianBlur(gray,5)
plt.imshow(median,cmap='gray')
```

```python
#avarage
av=cv2.blur(gray,(5,5))
plt.imshow(av,cmap='gray')
```
