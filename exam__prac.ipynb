```python
import numpy as np
min_val = np.min(img_gray)
max_val = np.max(img_gray)
contrast = ((img_gray-min_val)/(max_val-min_val)*255).astype(np.uint8)
plt.imshow(contrast,cmap='gray')
```

```python
smooth = cv2.GaussianBlur(img_gray,(5,5),0) #5, 5): the kernel size /// 0 sigmax
plt.imshow(smooth,cmap='gray')
```

```python
sharpen_kernal = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
sharpen =cv2.filter2D(img_gray,-1,sharpen_kernal)
plt.imshow(sharpen,cmap='gray')
```

```python
hist = cv2.calcHist([img_gray],[0],None,[256],[0,256])

plt.plot(hist)
plt.title("Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
```

```python
eq =cv2.equalizeHist(img_gray)
# Show histogram equalized image
plt.figure(figsize=(5, 5))
plt.imshow(eq, cmap='gray')
plt.title("Histogram Equalized")
plt.axis('off')
plt.show()
```
